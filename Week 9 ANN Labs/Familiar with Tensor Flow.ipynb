```python
import tensorflow as tf

# Check TensorFlow version
print("TensorFlow Version:", tf.__version__)

# Create a Tensor
tensor_a = tf.constant(5)
tensor_b = tf.constant(3)

# Perform Addition
tensor_c = tensor_a + tensor_b
print("Addition Result:", tensor_c.numpy())  # Convert to NumPy for printing
```

```python
import tensorflow as tf

# Create Tensors
tensor1 = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)
tensor2 = tf.Variable([[5, 6], [7, 8]], dtype=tf.float32)

# Reshape tensor1
reshaped_tensor = tf.reshape(tensor1, (4, 1))

# Perform element-wise multiplication
result = tensor1 * tensor2

print("Reshaped Tensor:\n", reshaped_tensor.numpy())
print("Multiplication Result:\n", result.numpy())
```
